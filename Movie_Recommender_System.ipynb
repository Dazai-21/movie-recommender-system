```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML Yt Projects Dataset/Movie Recommender System/tmdb_5000_movies.csv')
credits = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML Yt Projects Dataset/Movie Recommender System/tmdb_5000_credits.csv')
```

```python
movies.head(1)
```

```python
credits.head(1)
```

```python
movies = movies.merge(credits,on='title')
```

```python
movies.head(1)
```

```python
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.head(1)
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
import json
def convert(obj):
  L = []
  for i in json.loads(obj):
    L.append(i['name'])
  return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def convert3(obj):
  L = []
  counter = 0
  for i in json.loads(obj):
    if counter !=3:
      L.append(i['name'])
      counter+=1
    else:
      break
  return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
def fetch_director(obj):
  L = []
  for i in json.loads(obj):
    if i['job'] == 'Director':
      L.append(i['name'])
      break
  return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview'][0]
```

```python
movies['overview'] = movies['overview'].str.split()
```

```python
movies['genres'] = [ [i.replace(" ","") for i in x] for x in movies['genres'] ]
movies['keywords'] = [ [i.replace(" ","") for i in x] for x in movies['keywords'] ]
movies['cast'] = [ [i.replace(" ","") for i in x] for x in movies['cast'] ]
movies['crew'] = [ [i.replace(" ","") for i in x] for x in movies['crew'] ]
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head(1)
```

```python
movies = movies[['movie_id','title','tags']]
```

```python
movies['tags'] = movies['tags'].apply(' '.join)
```

```python
movies.head(1)
```

```python
movies['tags'] = movies['tags'].str.lower()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
  y = []
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
movies['tags'] = movies['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(movies['tags']).toarray()
```

```python
vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
import operator
sorted(enumerate(similarity[0]), reverse=True, key=operator.itemgetter(1))[1:6]
```

```python
def recommend(movie):
  movie_index = movies[movies['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(enumerate(distances), reverse=True, key=operator.itemgetter(1))[1:6]

  for i in movies_list:
    print(movies.iloc[i[0]].title)
```

```python
recommend('Spider-Man')
```
